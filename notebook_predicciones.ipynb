## 6.4. Modelo final para obtener predicciones

Realizamos la carga del modelo optimizado y del conjunto de datos de competición.

Finalmente, generamos las predicciones finales mediante el pipeline completo y exportadas al archivo predicciones.csv. Con este procedimiento, fue concluido el flujo de trabajo, dejando los resultados listos para la evaluación externa de la competición

```python
import pandas as pd
import joblib

# 1. Carga del modelo entrenado
# Cargamos el objeto .joblib que contiene tanto el modelo final optimizado como
#los pasos de preprocesamiento. Así evitamos el "data leakage" y aseguramos que los datos nuevos también se transformen
modelo_final = joblib.load('modelo_final.joblib')

# 2. Carga de los datos de competición 
# Datos que simulan nuevos cliente de los cuales no conocemos la variable objetivo
df_competicion = pd.read_pickle('bank_competition.pkl') 

# 3. Generación de predicciones
# El Pipeline preprocesará "df_competicion" antes de predecir
predicciones = modelo_final.predict(df_competicion)

# 4. Exportación a CSV
# Guardamos las predicciones con el formato esperado para la evaluación externa.
df_entrega = pd.DataFrame({'Prediccion': predicciones})
df_entrega.to_csv('predicciones.csv', index=False)

print("Proceso finalizado. El archivo 'predicciones.csv' ha sido generado.")
```
